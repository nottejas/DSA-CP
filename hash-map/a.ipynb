```python
# creating a class

class HashTable:
    def __init__(self):
        self.MAX = 100 # size of 100 elements
        self.arr = [None for i in range(self.MAX)] # list comprehension
        
    def get_hash(self, key):
        h = 0
        for char in key:
            h += ord(char)
        return h % self.MAX
    
    def __setitem__(self, key, value):
        h = self.get_hash(key)
        self.arr[h] = value
        
    def __getitem__(self, key):
        h = self.get_hash(key)
        return self.arr[h]
    
    def __delitem__(self, key):
        h = self.get_hash(key)
        self.arr[h] = None
```

```python
t = HashTable() # object    
t['march 8'] = 190
```
