```python
# creating a class

class HashTable:
    def __init__(self):
        self.MAX = 100 # size of 100 elements
        self.arr = [None for i in range(self.MAX)] # list comprehension
        
    def get_hash(self, key):
        h = 0
        for char in key:
            h += ord(char)
        return h % self.MAX
    
    def add(self, key, value):
        h = self.get_hash(key)
        self.arr[h] = value
        
    def get(self, key):
        h = self.get_hash(key)
        return self.arr[h]
```

```python
t = HashTable() # object    
t.add('march 8', 190)
```

```python
t.get('march 8')
```
