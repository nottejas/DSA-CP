```python
# creating a class

class HashTable:
    def __init__(self):
        self.MAX = 100 # size of 100 elements
        self.arr = [[] for i in range(self.MAX)] # now we store key-value pairs instead of jsu the value
        
    def get_hash(self, key):
        h = 0
        for char in key:
            h += ord(char)
        return h % self.MAX
    
    def __getitem__(self, key):
        h = self.get_hash(key)
        latest_value = None
        for element in self.arr[h]:
            if element[0] == key:
                latest_value = element[1]
        return latest_value

    
    def __setitem__(self, key, value):
        # h = self.get_hash(key)
        # found = False        
        # for idx, element in enumerate(self.arr[h]):
        #     if len(element) == 2 and element[0] == key:
        #         self.arr[h][idx] = (key, value) # modifying
        #         found=True
        #         break
        # if not found:
        #     self.arr[h].append((key, value)) # only happens if no data present
        h = self.get_hash(key)
        found = False        
        for idx, element in enumerate(self.arr[h]):
            if len(element) == 2 and element[0] == key:
                # Don't replace — instead just continue
                found = True
        # Always append new value
        self.arr[h].append((key, value))
    
    def __delitem__(self, key):
        h = self.get_hash(key)
        for idx, element in enumerate(self.arr[h]):
            if element[0] == key:
                del self.arr[h][idx]
```
